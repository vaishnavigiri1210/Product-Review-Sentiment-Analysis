```python
import pandas as pd     # data handling
```

```python
# data loading

df1 = pd.read_csv('../data/flipkart_product.csv', encoding='latin-1')
df2 = pd.read_csv('../data/Reviews.csv', encoding='utf-8-sig')
```

```python
# (number of rows, number of columns)
print("Shape df1:", df1.shape)
print("Shape df2:", df2.shape)
```

```python
df1.columns
```

```python
df2.columns
```

```python
df1.columns = df1.columns.str.replace('ï»¿', '').str.strip()
df2.columns = df2.columns.str.replace('ï»¿', '').str.strip()
```

```python
df1 = df1[['Review', 'rating']]
df2 = df2[['Review', 'rating']]
```

```python
df1.columns
```

```python
df2.columns
```

```python
# combine datasets
df = pd.concat([df1, df2], ignore_index=True)
```

```python
df.rename(columns={'Review': 'review_text'}, inplace=True)
```

```python
df.dropna(subset=['review_text', 'rating'], inplace=True)
```

```python
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
df.dropna(subset=['rating'], inplace=True)
```

```python
df['review_text'] = df['review_text'].astype(str).str.strip()
df = df[df['review_text'].str.len() > 1]
```

```python
df.drop_duplicates(subset=['review_text'], inplace=True)
```

```python
df = df.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
df.to_csv('../data/cleaned_data.csv', index=False, encoding='utf-8-sig')
```
